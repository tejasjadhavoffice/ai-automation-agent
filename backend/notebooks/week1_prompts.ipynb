# Week 1 Prompt Experiment

Goal: Compare zero-shot vs few-shot vs chain-of-thought for the same request.

## Shared Request

Use the exact same request text for all 3 runs.

```python
import subprocess
from pathlib import Path

backend_dir = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
request_text = "Read file sample.txt"

def run_style(style: str):
    command = [
        "python",
        "-m",
        "app.main",
        "--prompt-style",
        style,
        "--request",
        request_text,
    ]
    result = subprocess.run(command, cwd=backend_dir, capture_output=True, text=True)
    return result.stdout, result.stderr, result.returncode
```

```python
for style in ["zero-shot", "few-shot", "cot"]:
    out, err, code = run_style(style)
    print("=" * 60)
    print(style)
    print("return code:", code)
    print(out)
    if err:
        print("stderr:", err)
```
